Linear Regression : Age and Weight

```python
import pandas as pd
import numpy as np

# the board's table: age is the input, weight the output
age = np.array([[24], [25], [21], [27] ])  # 2D: one row per person, one column per feature
weight = np.array([64, 68, 60, 72])
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(age, weight)
```

```python
theta0 = model.intercept_
theta1 = model.coef_[0]

print(theta0, theta1)
```

```python
print(model.predict([[26]])[0])
```

```python
import matplotlib.pyplot as plt

x_line = np.linspace(age.min() - 1, age.max() + 1, 100).reshape(-1, 1)
y_line = model.predict(x_line)

plt.scatter(age, weight, color="blue", label="Data")
plt.plot(x_line, y_line, color="red",
         label=f"weight = {model.intercept_:.1f} + {model.coef_[0]:.2f} × age")
plt.xlabel("Age")
plt.ylabel("Weight")
plt.title("Linear regression: weight vs age")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("θ0 (intercept):", round(theta0, 2))
print("θ1 (slope):    ", round(theta1, 2))
print("hθ(26) from predict:", round(model.predict([[26]])[0], 2))
print("hθ(26) by hand:     ", round(theta0 + theta1 * 26, 2))
```
